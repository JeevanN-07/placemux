```python
import pandas as pd
import numpy as np
import hashlib

# 1. Simulate Raw Telemetry Data containing PII
np.random.seed(42)
raw_data = pd.DataFrame({
    "Student_Name": ["Rahul Sharma", "Ananya Rao", "Karthik V", "Priya Nair", "Sanjay Kumar"],
    "USN": [f"1BY22EC{i:03d}" for i in range(1, 6)],
    "Phone_Number": ["9876543210", "9876543211", "9876543212", "9876543213", "9876543214"],
    "DPDP_Consent_Granted": [True, True, False, True, False],
    "Department": ["ECE", "CSE", "ISE", "ECE", "ME"],
    "CTC_LPA": [14.2, 18.5, 12.0, 16.0, 9.5]
})

# 2. Apply DPDP Governance Rules (Data Masking & Consent Filtering)
def hash_pii(val):
    return hashlib.sha256(str(val).encode()).hexdigest()[:10]

# Mask PII fields for audit-compliant reporting
governed_df = raw_data.copy()
governed_df["Student_ID_Hashed"] = governed_df["USN"].apply(hash_pii)
governed_df["Student_Name"] = governed_df["Student_Name"].apply(lambda x: x[0] + "*** " + x.split()[-1][0] + "***")
governed_df["Phone_Number"] = "XXXXXXXXXX"

# Exclude non-consented data from downstream analytics
compliant_report = governed_df[governed_df["DPDP_Consent_Granted"] == True].drop(columns=["USN"])

# 3. Export Governance Compliance Artifact
compliant_report.to_csv("task_21_governance_applied_report.csv", index=False)
print("Task 21: DPDP Compliance Governance pipeline executed successfully!")
```
