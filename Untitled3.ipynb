```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.preprocessing import StandardScaler

# Set plot style
sns.set_theme(style="whitegrid")

# 1. Load Data
df = pd.read_csv('data/realistic_student_data.csv')

# Convert 'score' to numeric (forces any non-numeric strings to NaN)
df['score'] = pd.to_numeric(df['score'], errors='coerce')

# 2. Handle missing values (Fill NaN scores with median)
df['score'] = df['score'].fillna(df['score'].median())

# Select analysis columns (excluding identifiers and dates)
feature_cols = ['score']

# 3. Visual Distribution Review (Histograms & Box Plots)
for col in feature_cols:
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    
    # Histogram + KDE
    sns.histplot(df[col], kde=True, ax=axes[0], color='skyblue', bins=10)
    axes[0].set_title(f'Distribution: {col}')
    
    # Boxplot
    sns.boxplot(x=df[col], ax=axes[1], color='lightgreen')
    axes[1].set_title(f'Box Plot: {col}')
    
    plt.tight_layout()
    plt.show()

# 4. Central Tendency, Spread, and Normality Tests
review_summary = []

for col in feature_cols:
    data = df[col].dropna()
    mean_val = data.mean()
    std_val = data.std()
    median_val = data.median()
    iqr_val = data.quantile(0.75) - data.quantile(0.25)
    skew_val = data.skew()
    
    # Shapiro-Wilk Test for Normality
    stat, p_value = stats.shapiro(data)
    is_normal = p_value > 0.05

    review_summary.append({
        'Feature': col,
        'Mean': round(mean_val, 2),
        'Std Dev': round(std_val, 2),
        'Median': round(median_val, 2),
        'IQR': round(iqr_val, 2),
        'Skewness': round(skew_val, 2),
        'Shapiro P-Value': round(p_value, 4),
        'Normality Holds': is_normal
    })

summary_df = pd.DataFrame(review_summary)
print("=== Step 1-3: DISTRIBUTION REVIEW SUMMARY ===")
print(summary_df.to_string(index=False))

# 5. Transformation & Standardization
df_transformed = df[feature_cols].copy()

# Apply log transform if feature is heavily positively skewed (> 1.0)
for col in feature_cols:
    if df_transformed[col].skew() > 1.0 and (df_transformed[col] > 0).all():
        df_transformed[f'{col}_log'] = np.log1p(df_transformed[col])

# Standardize features for distance-based models/clustering
scaler = StandardScaler()
scaled_array = scaler.fit_transform(df_transformed)
df_scaled = pd.DataFrame(
    scaled_array, 
    columns=[f"{col}_scaled" for col in df_transformed.columns]
)

# Combine results with Student ID
final_dataset = pd.concat([df[['student_id']], df_scaled], axis=1)

print("\n=== Step 6: STANDARDIZED DATA FOR MODELLING ===")
print(final_dataset.head(10))
```
