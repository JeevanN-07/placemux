```python
%pip install seaborn matplotlib

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")

# 1. Load cleaned dataset from Task 4
df = pd.read_csv('data/cleaned_data.csv')

# 2. Distribution Plot (Key Numeric Variable)
plt.figure(figsize=(8, 4))
sns.histplot(df['score'], kde=True, color='skyblue', bins=5)
plt.title('Score Distribution Across Students', fontsize=12)
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.axvline(df['score'].mean(), color='red', linestyle='--', label=f"Mean Score ({df['score'].mean():.1f})")
plt.legend()
plt.tight_layout()
plt.show()

# 3. Categorical Comparison (Groupby Analysis)
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x='passed', y='score', palette='Set2', ci=None)
plt.title('Average Score by Pass Status', fontsize=12)
plt.xlabel('Passed Status')
plt.ylabel('Average Score')
plt.tight_layout()
plt.show()

# 4. Score Trend Across Submission Dates
plt.figure(figsize=(8, 4))
sns.lineplot(data=df, x='submission_date', y='score', marker='o', color='purple')
plt.title('Score Trend Over Submission Dates', fontsize=12)
plt.xlabel('Submission Date')
plt.ylabel('Score')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 5. Print Shortlist of Candidate Insights
print("=== EDA CANDIDATE INSIGHTS ===")
print(f"1. Score Distribution: Mean score is {df['score'].mean():.1f}, with a spread ranging from {df['score'].min()} to {df['score'].max()}.")
print(f"2. Pass Rate: {((df['passed'] == 'Yes').sum() / len(df)) * 100:.0f}% of students passed the evaluation.")
print("3. Follow-up Question: Why did submission scores drop significantly on 2026-07-02?")
```
