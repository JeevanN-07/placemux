```python
import pandas as pd
import numpy as np
from IPython.display import HTML, display

# 1. Dataset Construction
np.random.seed(42)
users = pd.DataFrame({
    'user_id': range(1, 51),
    'tier': np.random.choice(['Enterprise', 'Pro', 'Pay-per-App'], size=50, p=[0.4, 0.5, 0.1])
})

payments = pd.DataFrame({
    'user_id': np.random.choice(range(1, 51), size=200),
    'amount': np.random.choice([320, 190, 260], size=200),
    'status': np.random.choice(['success', 'failed', 'refunded'], size=200, p=[0.85, 0.1, 0.05])
})

# 2. Aggregations & Resilience Filtering
valid_payments = payments[payments['status'] == 'success']
summary = valid_payments.groupby('user_id')['amount'].sum().reset_index().merge(users, on='user_id')

dashboard_metrics = summary.groupby('tier').agg(
    Active_Users=('user_id', 'count'),
    Total_Revenue=('amount', 'sum'),
    ARPU=('amount', 'mean')
).reset_index()

dashboard_metrics['Total_Revenue'] = dashboard_metrics['Total_Revenue'].map('${:,.2f}'.format)
dashboard_metrics['ARPU'] = dashboard_metrics['ARPU'].map('${:,.2f}'.format)

# 3. Render HTML Dashboard Output
table_html = dashboard_metrics.to_html(index=False, classes='dash-table')

html_view = f"""
<style>
    .dash-box {{ font-family: sans-serif; border: 1px solid #cbd5e1; padding: 20px; border-radius: 8px; background: #ffffff; }}
    .dash-header {{ border-bottom: 2px solid #312e81; padding-bottom: 8px; margin-bottom: 16px; }}
    .dash-table {{ width: 100%; border-collapse: collapse; font-size: 14px; margin-top: 12px; }}
    .dash-table th {{ background: #312e81; color: white; padding: 10px; text-align: left; }}
    .dash-table td {{ padding: 10px; border-bottom: 1px solid #e2e8f0; }}
</style>
<div class="dash-box">
    <div class="dash-header">
        <h2 style="color:#1e1b4b; margin:0;">PlaceMux — Task 10 Revenue Dashboard</h2>
        <p style="color:#64748b; margin:4px 0 0 0;">Live Payment Analytics & Monetization Overview</p>
    </div>
    {table_html}
</div>
"""

display(HTML(html_view))
```
