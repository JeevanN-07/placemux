```python
import pandas as pd
import numpy as np
import hashlib

# 1. Simulate User Base for Experimentation (Both set to 100)
np.random.seed(42)
users = pd.DataFrame({
    "User_ID": [f"USR_{i:04d}" for i in range(1, 101)],
    "Department": np.random.choice(["CSE", "ISE", "ECE", "ME"], 100)
})

# 2. Deterministic Variant Assignment (A/B Split) using Hashing
def assign_variant(user_id):
    hash_val = int(hashlib.md5(user_id.encode()).hexdigest(), 16)
    return "Treatment" if hash_val % 2 == 0 else "Control"

users["Experiment_Variant"] = users["User_ID"].apply(assign_variant)

# 3. Simulate Engagement Metrics
users["Converted"] = np.where(
    users["Experiment_Variant"] == "Treatment",
    np.random.binomial(1, 0.65, len(users)),
    np.random.binomial(1, 0.50, len(users))
)

# 4. Summary & Evaluation
ab_summary = users.groupby("Experiment_Variant").agg(
    Total_Users=("User_ID", "count"),
    Conversions=("Converted", "sum"),
    Conversion_Rate=("Converted", "mean")
).reset_index()

print("A/B Experimentation Framework Results:")
print(ab_summary)

# 5. Export Experimentation Artifact
ab_summary.to_csv("task_22_experiment_report.csv", index=False)
print("\nExported task_22_experiment_report.csv successfully!")
```
