```python
import pandas as pd
import numpy as np
import io
import os

# 1. Ingest raw starter dataset
csv_data = """student_id,score,passed,submission_date
1,85 ,YES,2026-07-01
2, 42,no,2026-07-02
3,90,Yes,
4,65,YES,2026-07-04
5,50,NO,2026-07-05"""

df = pd.read_csv(io.StringIO(csv_data))

# Step 1: Type Coercion & String Standardization
df['score'] = df['score'].astype(int)
df['passed'] = df['passed'].astype(str).str.strip().str.capitalize()
df['submission_date'] = pd.to_datetime(df['submission_date'])

# Step 2: Missing Value Strategy (Fill null dates with forward fill)
df['submission_date'] = df['submission_date'].ffill()

# Step 3: Feature Engineering (Derive Grade column)
conditions = [
    (df['score'] >= 85),
    (df['score'] >= 70),
    (df['score'] >= 50)
]
choices = ['A', 'B', 'C']
df['grade'] = np.select(conditions, choices, default='F')

# Step 4: Ensure 'data' folder exists & Save Clean Output
os.makedirs('data', exist_ok=True)
df.to_csv('data/cleaned_data.csv', index=False)

print("--- Cleaned Dataset ---")
display(df)
print("\nCleaned dataset successfully saved to data/cleaned_data.csv!")
```
