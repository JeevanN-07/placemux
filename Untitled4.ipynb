```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Simulate Company-Side Funnel Events
funnel_events = [
    # Search events
    {"event_id": "e1", "event_type": "candidate_search", "recruiter_id": "rec_01", "timestamp": "2026-08-20T10:00:00Z"},
    {"event_id": "e2", "event_type": "candidate_search", "recruiter_id": "rec_01", "timestamp": "2026-08-20T10:05:00Z"},
    {"event_id": "e3", "event_type": "candidate_search", "recruiter_id": "rec_02", "timestamp": "2026-08-20T10:10:00Z"},
    {"event_id": "e4", "event_type": "candidate_search", "recruiter_id": "rec_03", "timestamp": "2026-08-20T10:12:00Z"},
    
    # Profile View events
    {"event_id": "e5", "event_type": "profile_viewed", "recruiter_id": "rec_01", "timestamp": "2026-08-20T10:06:00Z"},
    {"event_id": "e6", "event_type": "profile_viewed", "recruiter_id": "rec_02", "timestamp": "2026-08-20T10:15:00Z"},
    {"event_id": "e7", "event_type": "profile_viewed", "recruiter_id": "rec_03", "timestamp": "2026-08-20T10:18:00Z"},
    
    # Engagement / Outreach events
    {"event_id": "e8", "event_type": "outreach_sent", "recruiter_id": "rec_01", "timestamp": "2026-08-20T10:20:00Z"},
    {"event_id": "e9", "event_type": "outreach_sent", "recruiter_id": "rec_02", "timestamp": "2026-08-20T10:25:00Z"}
]

df_funnel = pd.DataFrame(funnel_events)

# 2. Compute Aggregated Funnel Stages
funnel_stages = df_funnel['event_type'].value_counts().reindex(
    ['candidate_search', 'profile_viewed', 'outreach_sent']
).reset_index()

funnel_stages.columns = ['Stage', 'Count']
funnel_stages['Conversion_Rate (%)'] = (funnel_stages['Count'] / funnel_stages['Count'].iloc[0] * 100).round(1)

print("--- Company Funnel Summary ---")
display(funnel_stages)

# 3. Visualize Company Funnel View
plt.figure(figsize=(7, 4))
bars = plt.bar(funnel_stages['Stage'], funnel_stages['Count'], color=['#312E81', '#4F46E5', '#10B981'], width=0.5)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.1, f"{int(yval)}", ha='center', va='bottom', fontweight='bold')

plt.title('Company-Side Funnel View (Search & Discovery)', fontweight='bold', pad=15)
plt.ylabel('Total Events', fontweight='bold')
plt.ylim(0, max(funnel_stages['Count']) + 2)
plt.show()
```
